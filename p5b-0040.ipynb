```python
# gastos diarios en un diccionario key dia gasto
# value lista de dias y lista de gastos
datos = {
    "mes": ["enero", "febrero", "Marzo", "abril", "mayo"],
    "Gasto": [5002, 3554, 8000, 4534, 6067]
}
# Dataframe tabla o marco de datos
df = pd.DataFrame(datos)
print("=== GASTOS SEMANALES ===")
print(df)

print("+-+-+-+-+-+-+jupyter Fabian-+-+-+-+-+-+-+")
print("+-+-+-+-+-+-+Hugo Fabian NC  0040-+-+-+-+-+-+")
```

```python
print("=== el total de gastos y el promedio de gastos ===")
print("Gasto total:", df["Gasto"].sum())
print("Gasto promedio:", df["Gasto"].mean())
print("\nDías con gasto mayor a $50:")
print(" Key Gasto values [5002, 3554, 8000, 4534, 6067]")
print(df[df["Gasto"] > 50])

print("+-+-+-+-+-+-+jupyter Fabian-+-+-+-+-+-+-+")
print("+-+-+-+-+-+-+Hugo Fabian NC  0040-+-+-+-+-+-+")
```
